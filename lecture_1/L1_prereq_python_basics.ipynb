# 一、Python变量基础

## 🔴 变量定义与赋值

**定义方式**：通过 `=` 直接赋值，无需提前声明类型  
**动态类型特性**：变量类型由赋值内容决定，可随时改变类型  
**多变量赋值**：支持同时为多个变量赋值，按顺序匹配右侧值

```python
# 定义方式
x = 10           # 整数赋值
name = "Alice"   # 字符串赋值
is_ok = True     # 布尔值赋值
print(x, name, is_ok)
```

```python
# 多变量赋值
a, b, c = 1, 2, 3
print(a, b, c)
x, y = 10, 20
x, y = y, x
print(x, y)
```

# 一、Python变量基础

## 🔴 变量作用域

**局部作用域（函数内）**  
- 在函数内部定义的变量仅在函数内可见，函数执行结束后销毁。

**全局作用域（函数外）**  
- 在模块顶层定义的变量可在全局范围内访问，包括函数内部（需用 `global` 声明修改）。

```python
# 局部作用域
def func():
    localVar = "局部变量"  # 仅在 func() 内可用
    print("函数内访问 localVar:", localVar)

func()
```

```python
# 下面这行会报错，因为 localVar 在函数外不可见
try:
    print(localVar)
except NameError as e:
    print("报错：", e)
```

```python
# 全局作用域
globalVar = "全局变量"

def func():
    global globalVar           # 声明全局变量
    print("函数内读取 globalVar:", globalVar)  # 现在可以正常访问
    globalVar = "修改后的全局变量"

func()
print("函数执行后 globalVar:", globalVar)
```

# 一、Python变量基础
## 🔴 变量作用域- LEGB 原则
**作用域查找规则（LEGB）**：
1. **Local（局部作用域）**  
   当前函数或代码块内定义的变量。
2. **Enclosing（嵌套函数的外层作用域）**  
   外层（但不是全局）的函数作用域变量。
3. **Global（全局作用域）**  
   模块顶层定义的变量。
4. **Built-in（内置作用域）**  
   Python 内置名称，如 `len`、`print` 等。

### 字符串操作

**索引与切片**

## 二、核心数据类型  

**2. 字符串（str）**  
### 字符串操作  
**常用方法**  
- `strip()`：去除首尾空格  
- `split()`：按空格分割  
- `replace(old, new)`：替换子串  

```python
s = "   Python is fun   "
print(s.strip())                  # 去除首尾空格: 'Python is fun'
print(s.split())                  # 按空格分割: ['Python', 'is', 'fun']
print(s.replace('fun', 'amazing'))  # 替换: 'Python is amazing'
```

## 二、核心数据类型

### 字符串操作  
**格式化输出**  
- 传统 `%` 格式化  
- `format()` 方法  
- Python 3.6+ 的 f-string 更简洁

```python
name = "Alice"
age = 30

print("My name is %s, age %d" % (name, age))         # 传统 % 格式化
print("My name is {}, age {}".format(name, age))     # format() 方法
print(f"My name is {name}, age {age}")               # f-string
```

## 二、核心数据类型  

**3. 列表（list）**

- **索引与切片**：与字符串类似，支持正向和负向索引（-1 表示最后一个元素）。

## 二、核心数据类型  

**3. 列表（list）**  

- **列表操作**  
- **增删改查**

```python
lst = [1, 2, 3]

lst.append(4)        # 尾部添加: [1, 2, 3, 4]
lst.insert(1, "a")   # 指定位置插入: [1, 'a', 2, 3, 4]
lst.remove("a")      # 删除指定值: [1, 2, 3, 4]
lst.pop(0)           # 删除指定索引元素（默认删除最后一个）: [2, 3, 4]
lst[0] = 100         # 修改元素: [100, 3, 4]
print(2 in lst)      # 检查元素是否存在: True
```

## 二、核心数据类型  

**3. 列表（list）**  

- **列表推导式**  
  简洁生成列表，替代循环操作。

```python
# 生成 1~10 的平方列表
squares = [x**2 for x in range(1, 11)]
# [1, 4, 9, ..., 100]

# 过滤偶数
evens = [x for x in range(20) if x % 2 == 0]
# [0, 2, 4, ..., 18]
```

- **元组特性**
  - **不可变性**：无法添加、删除或修改元素，适合存储固定数据（如坐标、置信息）。
  - **解包**：可将元组元素解压到多个变量。

## 二、核心数据类型

**5. 字典（dict）**

- **定义方式**：键值对 `{key: value}`  
  - 键必须唯一且不可变（如字符串、数字、元组）  
  - 值可以为任意类型

```python
# 定义字典
person = {
    "name": "Bob",
    "age": 35,
    "hobbies": ["reading", "gaming"]
}
```

## 二、核心数据类型

**5. 字典（dict）**

- **字典操作**
- **访问值**：通过键获取值  
- `get()` 方法可避免键不存在时报错

```python
person = {
    "name": "Bob",
    "age": 35,
    "hobbies": ["reading", "gaming"]
}

print(person["name"])                 # Bob
print(person.get("age"))               # 35
print(person.get("address", "未知"))   # 键不存在时返回默认值“未知”
```

## 二、核心数据类型

**5. 字典（dict）**

- **字典操作**
- **增删改查**

```python
person = {
    "name": "Bob",
    "age": 35,
    "hobbies": ["reading", "gaming"]
}

person["address"] = "New York"   # 添加新键值对
person["age"] = 36               # 修改值
del person["hobbies"]            # 删除键值对
print("name" in person)          # 检查键是否存在：True
```

## 二、核心数据类型

**5. 字典（dict）**

- **字典操作**
- **遍历字典**

```python
person = {
    "name": "Bob",
    "age": 35,
    "address": "New York"
}

# 遍历键
for key in person:
    print(key)
```

```python
# 遍历值
for value in person.values():
    print(value)
```

```python
# 遍历键值对
for key, value in person.items():
    print(key, ":", value)
```

## 6. 集合 (set)

- 集合操作
- 去重：交集、并集、差集、对称差集。

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a & b)       # 交集：{3, 4}
print(a | b)       # 并集：{1, 2, 3, 4, 5, 6}
print(a - b)       # 差集（a有b无）：{1, 2}
print(a ^ b)       # 对称差集（仅属于其中一个集合）：{1, 2, 5, 6} 
```

# 三、数据类型转换与内置函数

## 1. 类型转换函数

| 函数      | 作用                           | 示例 |
|-----------|--------------------------------|------|
| `int(x)`   | 将 `x` 转换为整数               | `int("123") → 123` |
| `float(x)` | 将 `x` 转换为浮点数             | `float("3.14") → 3.14` |
| `str(x)`   | 将 `x` 转换为字符串             | `str(100) → "100"` |
| `list(x)`  | 将可迭代对象转换为列表           | `list((1, 2, 3)) → [1, 2, 3]` |
| `tuple(x)` | 将可迭代对象转换为元组           | `tuple([4, 5, 6]) → (4, 5, 6)` |
| `dict(x)`  | 将键值对序列转换为字典           | `dict([("a", 1), ("b", 2)]) → {"a": 1, "b": 2}` |
| `set(x)`   | 将可迭代对象转换为集合           | `set("abc") → {'a', 'b', 'c'}` |

```python
## 示例：int(), float(), str()
print(int("123"))       # 123
print(float("3.14"))    # 3.14
print(str(100))         # "100"
print(list((1, 2, 3)))      # [1, 2, 3]
print(tuple([4, 5, 6]))     # (4, 5, 6)
print(dict([("a", 1), ("b", 2)]))   # {'a': 1, 'b': 2}
print(set("abc"))                   # {'a', 'b', 'c'}
```

# 三、数据类型转换与内置函数

## 2. 常用内置函数

### 序列操作
- `sorted(iterable, key=None, reverse=False)`  
  排序，返回新列表；`reverse=True` 表示降序，`key` 指定排序依据。

```python
# 示例 1：按字母升序排序
fruits = ["banana", "apple", "cherry"]
print(sorted(fruits))  # ['apple', 'banana', 'cherry']
```

```python
# 示例 2：按字母降序排序
fruits = ["banana", "apple", "cherry"]
print(sorted(fruits, reverse=True))  # ['cherry', 'banana', 'apple']
```

### 组合多个可迭代对象
- `zip(*iterables)`  
  将多个可迭代对象按元素打包成元组列表。

```python
# 示例 3：zip() 打包
names = ["Alice", "Bob"]
ages = [25, 30]
for name, age in zip(names, ages):
    print(name, age)  
# 输出：
# Alice 25
# Bob 30
```

# 五、总结与最佳实践

## 1. 变量命名规范
- 用小写字母和下划线组合（如 `user_name`），避免使用关键字（如 `list`、`dict`）
- 全局变量用大写字母（如 `MAX_SIZE`），类名用驼峰式（如 `UserProfile`）

---


## 2. 不可变类型优势
- 元组和字符串不可变，更适合作为字典键或多线程环境中的数据，避免意外修改

---

## 3. 选择合适数据结构
- 需要频繁增删用 **列表**
- 需要快速查找用 **字典**
- 需要唯一性用 **集合**

---

## 4. 避免性能陷阱
- 列表尾部添加元素 `append()` 效率高，**头部插入** `insert(0, x)` 效率低，可用 `collections.deque` 优化
- **字典、集合** 查找时间复杂度 O(1)，**列表** 查找 O(n)，大数据量优先选择前者

---

掌握 Python 变量与数据类型的核心特性，结合实际场景选择合适的数据结构，可高效解决文本处理、数据分析等问题，为自然语言处理等复杂任务打下基础。
